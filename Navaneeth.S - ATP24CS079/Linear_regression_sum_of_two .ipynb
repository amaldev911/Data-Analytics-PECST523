```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("sum_dataset.csv")
```

```python
#Just displaying the dataset
print(data.head())
```

```python
X = data[['A', 'B']]
y = data['Sum']
```

```python
#Splitting the data for training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)
```

```python
#calling model object
model = LinearRegression()

#training the model
model.fit(X_train, y_train)

#predciting??
predictions = model.predict(X_test)
```

```python
#comparing actual and predicted values
result = pd.DataFrame({'Actual' : y_test, 'Predicted' : predictions})
print(result.head(10))

```

```python
#Predicting results by taking input from the user
num1 = int(input("Enter the First Number : "))
num2 = int(input("Entter the Second Number : "))
predictions = model.predict([[num1, num2]])

print("Predicted Sum = ", predictions[0])
```
